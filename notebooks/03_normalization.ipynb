```python
import scanpy as sc
import sys
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
# Read adata_qc
adata_qc_path = PROJECT_ROOT / "data" / "interim" / "E-MTAB-8410_qc.h5ad"
adata_qc = sc.read_h5ad(adata_qc_path)
print(adata_qc.X.shape)

# Create a copy of qc'ed counts
adata_qc.layers["pre-normalization"] = adata_qc.X.copy()
```

```python
# Normalize so that each cell has a total of 10,000 count-like expression units
sc.pp.normalize_total(
    adata_qc,
    target_sum=10000,
)
adata_qc.layers["normalized_10k"] = adata_qc.X.copy()

# Log-transform normalized counts with a psuedocount of 1
sc.pp.log1p(adata_qc)
adata_qc.layers["normalized_log1p"] = adata_qc.X.copy()
```

```python
from src.ingest_data import write_h5ad
write_h5ad(
    adata_qc,
    output_path = PROJECT_ROOT / "data" / "interim" / "E-MTAB-8410_qc.h5ad"
)
```
